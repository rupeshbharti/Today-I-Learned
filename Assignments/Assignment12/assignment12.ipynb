```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("cleaned_avocado.csv")
df.head()
```

```python
plt.figure(figsize=(7,4))
sns.relplot(data=df, x="AveragePrice", y="Total Volume", hue="type")
plt.title("Total Volume by Average Price")
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data=df, x="AveragePrice", y="Total Volume", hue="type")
plt.title("Total Volume by Average Price")
plt.show()
```

## __Task 2__

```python
plt.figure(figsize=(8,4))
sns.lineplot(data=df, x="year", y="AveragePrice")
plt.title("Price By Year")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.lineplot(data=df, x="year", y="AveragePrice", marker="o")
plt.title("Price By Year")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.relplot(data=df, x="year", y="AveragePrice", kind="line", marker="o", col="type")
plt.title("Price By Year")
plt.show()
```

## __Task 3__

```python
sns.histplot(data=df, x="AveragePrice", bins=20)
plt.title("Average Price Distribution")
plt.xlabel("Average Price")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.kdeplot(data=df, x="AveragePrice", fill=True)
plt.title("Average Price Distribution")
plt.xlabel("Average Price")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.kdeplot(data=df, x="AveragePrice", hue="year", fill=True, color="ligtblue")
sns.rugplot(data=df, x="AveragePrice", hue="year", height=0.08)
plt.title("Average Price Distribution")
plt.xlabel("Average Price")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(data=df, x="AveragePrice", bins=20, kde=True)
plt.title("Average Price Distribution(Cobined Histplot and KDE)")
plt.xlabel("Average Price")
plt.ylabel("Frequency")
plt.show()
```

## __Task 4__

```python
sns.histplot(data=df, x="AveragePrice", y="year", cbar=True)
plt.title("Bivariate Hisplot")
```

```python
sns.kdeplot(data=df, x="AveragePrice", y="year", fill=True, thresh=0.05, cmap='mako')
plt.title("Bivariate KDE plot")
```

## __Task 5__

```python
sns.pairplot(df[['AveragePrice', 'Large Bags', 'Total Bags', 'type']], hue="type" )
plt.show()
```

```python
numeric_df = df[["AveragePrice","Total Volume", "Total Bags"]]
corr = numeric_df.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
