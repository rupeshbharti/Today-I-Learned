### Standard Scaler Formula ---> Z = (x - mu)/ sigma
__Mean__ : The formula subtracts the average value (mu) from every single data point. This process is called centering. By subtracting the center point from all values, the entire data distribution shifts so that its new balance point (the average) sits perfectly at 0.

__Standard Deviation__ : After centering the data, the formula divides every value by the original standard deviation (sigma). This process is called scaling. It compresses or expands the spread of the data points so that the average distance of all data points from the center (0) is exactly 1 unit.

## __Task 7__

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
```

```python
num_pipeline = Pipeline([
    ("scalar", StandardScaler())
])
```

```python
cate_pipeline = Pipeline([
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
```

```python
preprocessor = ColumnTransformer(transformers=[
    ("num", num_pipeline, numerical_features),
    ("cat", cate_pipeline, categorical_features)
])
```
