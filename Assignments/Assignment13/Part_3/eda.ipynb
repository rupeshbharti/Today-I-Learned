# __Exploratory Data Analysis__

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("Tatanic.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df = df.drop(columns=['Cabin'])
```

```python
df.head()
```
