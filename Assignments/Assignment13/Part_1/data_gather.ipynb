## __Task 3__ --> Loading data form SQL

```python
import sqlite3
import pandas as pd

# Connect to database (creates sample.db if it doesn't exist)
conn = sqlite3.connect('sample.db')
cursor = conn.cursor()

# Ensure table exists from previous step
cursor.execute('''
    CREATE TABLE IF NOT EXISTS employees (
        id TEXT PRIMARY KEY,
        name TEXT,
        department TEXT,
        salary INTEGER
    )
''')

# Step 3: Insert at least 5 records
records_to_insert = [
    ("EMP101", "Arjun Sharma", "Engineering", 85000),
    ("EMP102", "Priya Patel", "Design", 72000),
    ("EMP103", "Rohan Das", "Engineering", 78000),
    ("EMP104", "Ananya Iyer", "Marketing", 65000),
    ("EMP105", "Kabir Mehta", "HR", 60000)
]

cursor.executemany('''
    INSERT OR IGNORE INTO employees (id, name, department, salary)
    VALUES (?, ?, ?, ?)
''', records_to_insert)
conn.commit()

# Step 4: Read the table into Pandas using SQL query
query = "SELECT * FROM employees"
df = pd.read_sql_query(query, conn)

# Close the database connection
conn.close()

# Step 5: Display the DataFrame
print("--- Final Employees DataFrame ---")
print(df)
```
