## __Task 1__ : 

```python
class Product():

    def __init__(self, name, price, category):
        self.name = name
        self.__price = price
        self.category = category

    def get_info(self):
        return f"Product name is {self.name}, price is {self.__price}, category is {self.category}"

## Task 2
    def get_price(self):
        print(f"Price {self.__price}")

    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
            print(f"Price set to : {new_price}")
        else:
            print("Error.")
```

```python
product1 = Product("Paneer",400,"Grocery")
```

```python
product2 = Product("Apple",100,"Fruit")
```

```python
product1.get_info()
product2.get_info()
```

```python
class ElectronicProduct(Product):

    def __init__(self, name, price, category, waranty):
        super().__init__(name, price, category)
        self.waranty = waranty

    def get_info(self):
        parent_info = super().get_info()
        return f"{parent_info} waranty is {self.waranty} year"
        
```

```python
prod3 = ElectronicProduct("Earbuds",799,"Electronic",1)
```

```python
prod3.get_info()
```

## __Task 4__

```python
class Laptop(Product):

    def __init__(self, name, price, category, batery):
        super().__init__(name, price, category)
        self.batery = batery

    def get_info(self):
        parent_info = super().get_info()
        return f"{parent_info} Batery capacity {self.batery}"
```

```python
class Mobile(Product):

    def __init__(self, name, price, category, ram):
        super().__init__(name, price, category)
        self.ram = ram

    def get_info(self):
        parent_info = super().get_info()
        return f"{parent_info} Ram {self.ram}"
```

```python
product_list = [
    Laptop("Dell",40000,"Electric",4000),
    Laptop("Lenovo",50000,"Electric",5000),
    Mobile("Samsung",30000,"Electric",8),
    Mobile("Redmi",25000,"Electric",12)
]
```

```python
for product in product_list:
    print(product.get_info())
```
