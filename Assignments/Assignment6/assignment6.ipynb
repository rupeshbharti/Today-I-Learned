## __Task 1__

```python

try:
    numerator_input = input("Enter the numerator: ").strip()
    denominator_input = input("Enter the denominator: ").strip()
    
    numerator = float(numerator_input)
    denominator = float(denominator_input)
    
    result = numerator / denominator

except ValueError:
    print("Error: Invalid input. Please enter a valid number.")

except ZeroDivisionError:
    print("Error: Cannot divide by zero.")

else:
    print(f"Result: {result}")

finally:
    print("Operation Complete")
```

## __Task 2__

```python
class NegativePriceError(Exception):
    """Exception raised for negative price values."""
    pass

prices = [120, 350, 'abc', 500, -200, 800]
total = 0

print("--- Processing Bill ---")

for item in prices:
    try:
        if item < 0:
            raise NegativePriceError("Negative price not allowed")
            
        total += item
        
    except TypeError:
        print(f"Skipped item '{item}': Unsupported operand type(s) for +")
        
    except NegativePriceError as npe:
        print(f"Skipped item '{item}': {npe}")

print(f"\nFinal Bill Total: {total}")
```

## __Task 3__

```python
def check_age(age):
    if age < 1 or age > 120:
        raise ValueError("Age must be between 1 and 120")
    return True

try:
    age_value = int(input("Enter yur age:"))
    
    check_age(age_value)
    print(f"Success! Your registered age is: {age_value}")

except ValueError as error:
    print(f"Error occurred: {error}")
```

## __Task 4__

```python
try:
    filename = input("Enter the filename to open: ").strip()

    with open(filename, "r") as file:
        print("\n--- File Content (First 3 Lines) ---")
        
        for i in range(3):
            line = file.readline()
            if not line:
                break  
            print(line, end="")

except FileNotFoundError:
    print("Error: The file does not exist. Please check the spelling.")

except PermissionError:
    print("Error: You do not have permission to read this file.")

finally:
    print("\nFile operation attempted.")
```

## __Task 5__

```python
cart = []

print("--- Safe Shopping Cart System ---")
print("Enter item prices. Type 'q' to finish shopping.\n")

while True:
    user_input = input("Enter price of the item: ").strip().lower()

    if user_input == 'q':
        break
        
    try:
        price = float(user_input)
        
        if price < 0:
            raise ValueError("Price cannot be negative")
        
        cart.append(price)
    
    except ValueError as error:
        if str(error) == "Price cannot be negative":
            print(f"Error: {error}. Item skipped.")
        else:
            print("Error: Invalid numeric input. Please enter a valid price or 'q'.")

print("\n--- Order Summary ---")
print(f"Total items: {len(cart)}")
print(f"Total bill:  ₹{sum(cart):.2f}")
```
