## __Task 1__

```python

try:
    # 1. Takes two inputs from the user: numerator and denominator
    numerator_input = input("Enter the numerator: ").strip()
    denominator_input = input("Enter the denominator: ").strip()
    
    # 2. ValueError will be handled if input is not a number
    numerator = float(numerator_input)
    denominator = float(denominator_input)
    
    # ZeroDivisionError will be handled if denominator = 0
    result = numerator / denominator

except ValueError:
    # Handle non-numeric inputs gracefully
    print("Error: Invalid input. Please enter a valid number.")

except ZeroDivisionError:
    # Handle division by zero constraints
    print("Error: Cannot divide by zero.")

else:
    # 3. If no error occurs, print the result inside the else block
    print(f"Result: {result}")

finally:
    # 4. In the finally block, print "Operation Complete"
    print("Operation Complete")
```

## __Task 2__

```python

prices = [120, 350, 'abc', 500, -200, 800]
total = 0

print("--- Processing Bill ---")

# 1. Iterates through the list
for item in prices:
    try:
        # Check if the type is numeric (int or float)
        # If it's a string like 'abc', adding it to an int triggers a TypeError
        if not isinstance(item, (int, float)):
            raise TypeError("Value is not a number")
            
        # 3. Custom exception check for negative values
        if item < 0:
            raise ValueError("Negative price not allowed")
            
        # 2. Tries to add only valid (positive numerical) prices to the total
        total += item
        
    except TypeError as te:
        print(f"Skipped item '{item}': {te}")
        
    except ValueError as ve:
        print(f"Skipped item '{item}': {ve}")

# 4. Prints the running total / final total
print(f"\nFinal Bill Total: {total}")
```

## __Task 3__

```python
# 1. Write a function check_age(age)
def check_age(age):
    # Raise a custom ValueError if age is out of the required range
    if age < 1 or age > 120:
        raise ValueError("Age must be between 1 and 120")
    return True


# 2. Main code execution block
try:
    # Take age input from the user
    user_input = input("Enter your age: ").strip()
    
    # Convert input to integer (handles potential non-numeric inputs)
    age_value = int(user_input)
    
    # Validate the converted age value
    check_age(age_value)
    
    # Success message if validation succeeds
    print(f"Success! Your registered age is: {age_value}")

except ValueError as error:
    # Catch and print the custom error message or invalid input formatting
    print(f"Error occurred: {error}")
```

## __Task 4__

```python
try:
    # 1. Ask the user for a filename
    filename = input("Enter the filename to open: ").strip()
    
    # 2. Try to open and read the file
    with open(filename, "r") as file:
        print("\n--- File Content (First 3 Lines) ---")
        
        # 4. If successful, print first 3 lines of the file
        for i in range(3):
            line = file.readline()
            if not line:
                break  # Stop if the file has fewer than 3 lines
            print(line, end="")

# 3. Handle specific exceptions
except FileNotFoundError:
    print("Error: The file does not exist. Please check the spelling.")

except PermissionError:
    print("Error: You do not have permission to read this file.")

# 5. Use finally to print the final execution tracker
finally:
    print("\nFile operation attempted.")
```

## __Task 5__

```python
# 1. Has a cart list
cart = []

print("--- Safe Shopping Cart System ---")
print("Enter item prices. Type 'q' to finish shopping.\n")

# 2. Runs a loop asking user to enter prices
while True:
    user_input = input("Enter price of the item: ").strip().lower()
    
    # 3. Stops when user enters 'q'
    if user_input == 'q':
        break
        
    # 4. Inside the loop execution block
    try:
        # Convert input to a float
        price = float(user_input)
        
        # Raise custom exception if price is negative
        if price < 0:
            raise ValueError("Price cannot be negative")
            
        # Append valid price to the cart
        cart.append(price)
        
    # Handle ValueError if user enters invalid input or negative price
    except ValueError as error:
        # Check if the error message is our custom negative condition
        if str(error) == "Price cannot be negative":
            print(f"Error: {error}. Item skipped.")
        else:
            print("Error: Invalid numeric input. Please enter a valid price or 'q'.")

# 5. At the end, print calculations
print("\n--- Order Summary ---")
print(f"Total items: {len(cart)}")
print(f"Total bill:  ${sum(cart):.2f}")
```
