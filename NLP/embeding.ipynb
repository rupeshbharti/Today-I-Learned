# __Word Embedding__

```python
import pandas as pd
import re
import string
```

```python
df = pd.read_csv("movie_reviews.csv", delimiter="\t")
df.head()
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'https?://\S+', '', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = text.encode('ascii', 'ignore').decode('ascii')
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
df['clean_review'] = df['review'].apply(clean_text)
df[['review', 'clean_review']].head()
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
df['token'] = df['clean_review'].apply(word_tokenize)
```

```python
df['token_count'] = df['token'].apply((len))
```

```python
df.head()
```

```python
from nltk.corpus import stopwords
from string import punctuation
```

```python
stop = stopwords.words("english")
```

```python
stop_words = list(punctuation) + stop
```

```python
df['filtered_token'] = df['token'].apply(
    lambda token: [word for word in token if word not in stop_words]
)
```

```python
df.head()
```

```python
custom_stop_words = set(stopwords.words("english"))

custom_stop_words.discard("not")
custom_stop_words.discard("no")
custom_stop_words.discard("never")
```

```python
df['filtered_token'] = df['token'].apply(
    lambda token: [word for word in token if word not in custom_stop_words]
)
```

```python
df.head()
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
stem = PorterStemmer()
lema = WordNetLemmatizer()
```

```python
df['stemmed_token'] = df['filtered_token'].apply(
    lambda token: [stem.stem(word) for word in token]
)
```

```python
df['lematize_token'] = df['filtered_token'].apply(
    lambda token: [lema.lemmatize(word, pos='v') for word in token]
)
```

```python
df.head()
```

```python
df['processed_token'] = df['lematize_token'].apply(
    lambda token: " ".join(token)
)

df.head()
```

```python
df.to_csv("processed.csv", index=False)
```
