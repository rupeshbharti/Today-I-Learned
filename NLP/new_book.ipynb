```python
import joblib
```

```python
loaded_model = joblib.load("sentiment_model.pkl")
```

```python
load_vector1 = joblib.load("tfidf_vectors.pkl")
```

```python
load_vector2 = joblib.load("BoW_vectors.pkl")
```

```python
new_review = [
    "The movie was absolutely fantastic"
]
```

```python
new_review2 = [
    "The movie was absolutely Boring"
]
```

```python
review_vector1 = load_vector1.transform(new_review2)
```

```python
review_vector2 = load_vector2.transform(new_review)
```

```python
print(hasattr(loaded_model, "classes_"))
```

```python
prediction = loaded_model.predict(review_vector2)
print(prediction)
```

```python
prediction = loaded_model.predict(review_vector1)
print(prediction)
```
