```python
def clean_text(tex):
    tex = tex.lower()
    tex = re.sub(r'https?://\S+', '', tex)
    tex = tex.translate(str.maketrans('', '', string.punctuation))
    tex = tex.encode('ascii', 'ignore').decode('ascii')
    tex = re.sub(r'\s+', ' ', tex).strip()
    return tex
```
